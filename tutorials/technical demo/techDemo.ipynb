## Frontend Communication with the cubing engine : Recipe

The cubing engine has been constructed to operate on 2 different levels

1) **Backend**, which is aimed at software developers who wish to extend the framework
    * **Geospatial processing methods**
    * **ETL pipeline** 
    * **Dataset adaptors**
2) **Frontend**, which is aimed at users who communicate their data needs through the use of a recipe

Engine functions through the use of a **recipe** which is passed to the orchestrator. This abstracts away much of the complexity and reduces the engine use to **recipe** construction. The end script that the user would have to submit is reduced to the following

## Functionality of the cubing engine

The engine supports geospatial methods that

1) Align data from disparate sources to a shared master grid
    * Reprojection to shared CRS
    * Upsampling or downsampling to shared resolution
2) Vector processing
3) STAC builder 

Alignment of the data and construction of a STAC entry of the end product happens automatically during the processing step of the cube and is encoded in the cube classes and the dataset specific adaptors. Handling vector data is where the end user has more freedom

## Vector data

Currently the vector data code is constructed with GBIF data in mind and as such is (currently) more constructed at tackling issues that are inherent to the dataset itself. In the future when additional vector datasets are added we will provide more generalized frameworks.

GBIF data handling deals with **spatially uncertain point data* for which we provide

1) **Geometry conversion** of tabular Darwincore data
    * Point geometry
    * Point cloud geometry
    * Polygonal geometry
2) Point to point cloud conversion through **Monte Carlo sampling**
3) Geometry to grid cell
    * **Classification**
    * **Weight fraction** 

### Monte Carlo sampling of spatially uncertain vector data

Due to the tabular nature of the GBIF data we have access to data that is characterized by 

* decimalLongitude and decimalLatitude
* coordinateUncertaintyInMeters

These individual records can have an uncertainty that exceeds the resolution of the grid on which the data is aligned. As such the cubing engine offers a method to randomize data through a **Monte Carlo point resampling**. 

In this method we construct a probability distribution which can take the shape of 

* Normal distribution with 3 $\sigma $ spanning the uncertainty radius
* Uniform distribution spanning a circle with the radius equal to the uncertainty radius

Using this distribution we sample a specified amount of points and utilize the result for either classification/weight fraction calculation

#### Visualization function random sampling

```python
import matplotlib.pyplot as plt
import numpy as np
import geopandas as gpd
from shapely.geometry import Point
from bmc.engine.vector import vectorEngine

engine = vectorEngine()

def MC_sampling_plot():
    # Setup the Scenarios
    # We will use a localized metric projection (EPSG:3035) to make distances clear.
    # Let's pretend we have a 1000m x 1000m grid.
    point_center = Point(500, 500)    # Scenario A: Center of the cell

    gdf = gpd.GeoDataFrame({
        'id': ['A_Center'],
        'coordinateuncertaintyinmeters': [500.0] # Increased to 500m uncertainty
    }, geometry=[point_center], crs="EPSG:3035")

    # Generate the Point Clouds (500 passes)
    # Uniform Distribution
    uniform_gdf = engine.generate_spatial_point_clouds(
        gdf=gdf,
        n_passes=500, # Increased passes to clearly show density
        uncertainty_col="coordinateuncertaintyinmeters",
        output_col="point_cloud",
        distribution="uniform",
        random_state=42 # Set seed for reproducible plots
    )

    # Gaussian Distribution
    gaussian_gdf = engine.generate_spatial_point_clouds(
        gdf=gdf,
        n_passes=500,
        uncertainty_col="coordinateuncertaintyinmeters",
        output_col="point_cloud",
        distribution="gaussian",
        random_state=42
    )

    # Visualization Setup
    fig, axes = plt.subplots(1, 2, figsize=(14, 7))
    plt.suptitle("Probabilistic Spatial Point Clouds (500 Passes | 500m Uncertainty)", fontsize=16, fontweight='bold')

    def plot_cloud(ax, multipoint, color):
        """Helper to extract and plot individual points from the MultiPoint object."""
        x = [p.x for p in multipoint.geoms]
        y = [p.y for p in multipoint.geoms]
        # Removed edge colors and lowered alpha to visualize dense overlaps better
        ax.scatter(x, y, color=color, alpha=0.3, s=15, edgecolor='none')

    # Extract uncertainty radius from the geodataframe to keep it dynamic
    uncertainty_radius = gdf.loc[0, 'coordinateuncertaintyinmeters']

    # Setup Grid Lines to illustrate the cell boundaries
    for ax in axes:
        ax.set_aspect('equal')
        # Adjusted limits to center the 1km box beautifully
        ax.set_xlim(-200, 1200)
        ax.set_ylim(-200, 1200)
        
        # Emphasize the 1km grid cell boundaries
        ax.axvline(x=0, color='black', linestyle='--', alpha=0.6, linewidth=1.5)
        ax.axvline(x=1000, color='black', linestyle='--', alpha=0.6, linewidth=1.5)
        ax.axhline(y=0, color='black', linestyle='--', alpha=0.6, linewidth=1.5)
        ax.axhline(y=1000, color='black', linestyle='--', alpha=0.6, linewidth=1.5)
        
        # Add a standard coordinate grid for better distance readability
        ax.grid(True, linestyle=':', alpha=0.7)
        ax.set_xticks(np.arange(-250, 1251, 250))
        ax.set_yticks(np.arange(-250, 1251, 250))
        
        ax.set_xlabel("Easting (m)")
        ax.set_ylabel("Northing (m)")
        
        # NEW: Draw the uncertainty circle around the centroid
        circle = plt.Circle((point_center.x, point_center.y), uncertainty_radius, 
                            color='red', fill=False, linestyle='-', linewidth=2, 
                            alpha=0.8, label=f'Uncertainty Radius ({int(uncertainty_radius)}m)')
        ax.add_patch(circle)

    # PANEL 1: Uniform Distribution
    axes[0].set_title("1. Uniform Distribution\nEqual probability across the entire radius", fontweight='bold')
    plot_cloud(axes[0], uniform_gdf.loc[0, 'point_cloud'], 'forestgreen') # Center
    # Plot original centroid
    gdf.plot(ax=axes[0], color='black', marker='X', markersize=100, zorder=5, label='Original Centroid')
    axes[0].legend(loc='upper right')

    # PANEL 2: Gaussian Distribution
    axes[1].set_title("2. Gaussian Distribution\nProbability clusters heavily around centroid", fontweight='bold')
    plot_cloud(axes[1], gaussian_gdf.loc[0, 'point_cloud'], 'forestgreen') # Center
    # Plot original centroid
    gdf.plot(ax=axes[1], color='black', marker='X', markersize=100, zorder=5, label='Original Centroid')
    axes[1].legend(loc='upper right')

    plt.tight_layout()
    plt.show()
```

### Geometry conversion

Due to the tabular nature of the GBIF data we have access to data that is characterized by 

* decimalLongitude and decimalLatitude
* coordinateUncertaintyInMeters

Using this information we can construct 3 different types of geometry depending on the user's specifications

1) Point data (keep uncertainty as a feature separate from geometry)
2) Point cloud based on the centroid with the uncertainty incorporated
3) Circular polygon based on the centroid with the uncertainty incorporated

#### Geometry conversion visualization code

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np

def GC_visualization():
    # Define a single record to trace across all geometry types
    raw_data = {
        'obs_id': ['Single Record (4-Way Corner)'],
        'x': [1000],
        'y': [1000],
        'uncertainty': [500.0]
    }
    df = pd.DataFrame(raw_data)

    # A. Exact Points
    points_gdf = engine.coordinate_to_geometry(
        df, x_col='x', y_col='y', output_type='point', input_crs='EPSG:3035'
    )

    # B. Buffered Polygons
    polygons_gdf = engine.coordinate_to_geometry(
        df, x_col='x', y_col='y', uncert_col='uncertainty', 
        output_type='polygon', input_crs='EPSG:3035', quad_segs=16
    )

    # C. Uniform Point Cloud (200 passes)
    clouds_uni_gdf = engine.coordinate_to_geometry(
        df, x_col='x', y_col='y', uncert_col='uncertainty', 
        output_type='point_cloud', input_crs='EPSG:3035',
        point_cloud_config={'distribution': 'uniform', 'n_passes': 200, 'random_state': 43}
    )

    # D. Gaussian Point Cloud (200 passes)
    clouds_gauss_gdf = engine.coordinate_to_geometry(
        df, x_col='x', y_col='y', uncert_col='uncertainty', 
        output_type='point_cloud', input_crs='EPSG:3035',
        point_cloud_config={'distribution': 'gaussian', 'n_passes': 200, 'random_state': 43}
    )

    fig, axes = plt.subplots(2, 2, figsize=(16, 16))
    plt.suptitle("Geometry Conversions for a Single Record (500m Uncertainty)", fontsize=20, fontweight='bold', y=0.95)

    # Use one consistent color for the single record
    plot_color = 'rebeccapurple'

    def draw_master_grid(ax):
        """Draws a mock 2x2 grid of 1000m cells for visual mapping context."""
        ax.set_aspect('equal')
        ax.set_xlim(-200, 2200)
        ax.set_ylim(-200, 2200)
        for x in [0, 1000, 2000]:
            ax.axvline(x=x, color='black', linestyle='-', alpha=0.8, linewidth=2)
        for y in [0, 1000, 2000]:
            ax.axhline(y=y, color='black', linestyle='-', alpha=0.8, linewidth=2)
        ax.grid(True, linestyle=':', alpha=0.5)
        ax.set_xticks(np.arange(0, 2001, 250))
        ax.set_yticks(np.arange(0, 2001, 250))
        ax.set_xlabel("Easting (m)")
        ax.set_ylabel("Northing (m)")

    def plot_cloud(ax, multipoint, color):
        """Helper to extract and plot individual points from a MultiPoint object."""
        x_coords = [p.x for p in multipoint.geoms]
        y_coords = [p.y for p in multipoint.geoms]
        ax.scatter(x_coords, y_coords, color=color, alpha=0.5, s=15, edgecolor='none', zorder=3)

    # Extract the base point geometry once to reuse the centroid in all plots
    base_point = points_gdf.iloc[0].geometry

    # PANEL A: Exact Points
    ax = axes[0, 0]
    draw_master_grid(ax)
    ax.set_title("1. output_type = 'point'\n1D Coordinate (No Uncertainty)", fontweight='bold')
    ax.plot(base_point.x, base_point.y, marker='X', color=plot_color, markersize=14, label=points_gdf.iloc[0]['obs_id'], markeredgecolor='white', zorder=5)
    ax.legend(loc='upper right', framealpha=0.9)

    # PANEL B: Buffered Polygons
    ax = axes[0, 1]
    draw_master_grid(ax)
    ax.set_title("2. output_type = 'polygon'\nRigid Buffer (Equal probability across area)", fontweight='bold')
    # Plot the polygon area
    gpd.GeoSeries([polygons_gdf.iloc[0].geometry]).plot(ax=ax, color=plot_color, alpha=0.4, edgecolor='black', linewidth=1, zorder=3)
    # Plot original centroid for reference
    ax.plot(base_point.x, base_point.y, marker='X', color=plot_color, markersize=10, markeredgecolor='white', zorder=5)

    # PANEL C: Uniform Point Cloud
    ax = axes[1, 0]
    draw_master_grid(ax)
    ax.set_title("3. output_type = 'point_cloud' (Uniform)\nProbabilistic points scattered evenly in radius", fontweight='bold')
    plot_cloud(ax, clouds_uni_gdf.iloc[0]['point_cloud'], plot_color)
    ax.plot(base_point.x, base_point.y, marker='X', color='black', markersize=8, zorder=5)

    # PANEL D: Gaussian Point Cloud
    ax = axes[1, 1]
    draw_master_grid(ax)
    ax.set_title("4. output_type = 'point_cloud' (Gaussian)\nProbabilistic points clustered near origin centroid", fontweight='bold')
    plot_cloud(ax, clouds_gauss_gdf.iloc[0]['point_cloud'], plot_color)
    ax.plot(base_point.x, base_point.y, marker='X', color='black', markersize=8, zorder=5)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
```

### Point classification and fractional weight calculation

When dealing with a geometry type that is different from a single point, the engine needs to be directed as to what needs to be done with this data. Point clouds and polygons can span multiple cells and can thus be processed in 2 ways:

1) **Classification** (only applies to point clouds), using 2 different methods we can mathematically compute to which cell the geometry most likely belongs.
2) **Fractional weight calculation**, we can determine how much of a record falls in each cell it is present thus resulting in a fractional weight that sums up to 1.

For these methods we can use 2 different algorithms to obtain the end result:

1) **kdtree** (only applies to classification), compute which centroid minimizes the distance to the point cloud and the record gets assigned to this cell.

<img src="img/kdtree%20classification.png" alt="kdtree classification" width="75%">

2) **intersect**
   - In classification we assign the record to the cell that contains the majority of the randomly sampled points.
   - In fractional weight calculation, we determine either the ratio of points in the cell compared to the total amount of points or the area of the total polygonal surface.

<img src="img/datatree.png" alt="Standaard output" width="50%">

```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import geopandas as gpd
import rioxarray

# 1. Fetch European NUTS regions from Eurostat (GISCO)
url = "https://gisco-services.ec.europa.eu/distribution/v2/nuts/geojson/NUTS_RG_10M_2021_4326.geojson"
nuts = gpd.read_file(url)

# Make sure xarray knows the CRS (using EPSG:3035)
monthly_data = monthly_data.rio.write_crs("EPSG:3035") 

# Filter specifically for Flanders (NUTS_ID: BE2) and project to match the raster
flanders = nuts[nuts['NUTS_ID'] == 'BE2'].to_crs(monthly_data.rio.crs)

# 2. Clip the ENTIRE dataset to Flanders BEFORE computing statistics.
# drop=True crops the spatial extent tightly to the bounding box, making the map larger.
masked_data = monthly_data.rio.clip(flanders.geometry, flanders.crs, drop=True)

# 3. Aggregate spatially to create a 1D time series
# (xarray automatically ignores the NaN values outside the border)
tas_ts = masked_data['tas'].mean(dim=['x', 'y'])

monthly_grouped = tas_ts.groupby('time.month')
monthly_max = monthly_grouped.max()
monthly_min = monthly_grouped.min()
monthly_mean = monthly_grouped.mean()

# 4. Select a single time slice for the spatial raster plot
time_index = 0 
single_slice = masked_data['tas'].isel(time=time_index)
slice_date = pd.to_datetime(single_slice.time.values).strftime('%Y-%m')

# =========================================================
# FIGURE SETUP: 1x2 Grid
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# --- PANEL 1: Monthly Extremes & Average (1D) ---
ax1 = axes[0]
months = np.arange(1, 13)
month_labels = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
                'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

ax1.fill_between(months, monthly_min, monthly_max, color='lightblue', alpha=0.5, 
                 label='Historical Min-Max Range (Flanders)')
ax1.plot(months, monthly_mean, color='red', marker='o', linewidth=2, label='Monthly Average')

ax1.set_xticks(months)
ax1.set_xticklabels(month_labels)
ax1.set_xlabel('Month')
ax1.set_ylabel('Temperature (tas)')
ax1.set_title('Monthly Temperature Extremes (Flanders: 2004-2021)')
ax1.legend()
ax1.grid(True, linestyle=':', alpha=0.7)

# --- PANEL 2: Spatial Raster Map (Masked to Flanders) ---
ax2 = axes[1]
plot = single_slice.plot(
    ax=ax2, 
    cmap='coolwarm', 
    add_colorbar=True,
    cbar_kwargs={'label': 'Temperature (tas)'}
)

# Overlay the border of Flanders for visual clarity
flanders.plot(ax=ax2, facecolor="none", edgecolor="black", linewidth=1.5)

ax2.set_aspect('equal', 'datalim') 
ax2.set_title(f'Spatial Distribution (Flanders): {slice_date}')
ax2.set_xlabel('Easting (x)')
ax2.set_ylabel('Northing (y)')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

# 1. Load Data & Geometries
dataset_dir = "tech_demo_output/techDemo_cube/gbif"
aggregated_path = f"{dataset_dir}/gbif_aggregated.parquet"
dimension_table_path = f"{dataset_dir}/gbif_EEA_1km_dimension_table.parquet"

gbif_agg = pd.read_parquet(aggregated_path)
grid_geoms = gpd.read_parquet(dimension_table_path)

# Merge tabular data with spatial geometries
gbif_spatial = grid_geoms.merge(gbif_agg, on='grid_idx', how='inner')

# 2. Fetch & Prepare Flanders Boundary
url = "https://gisco-services.ec.europa.eu/distribution/v2/nuts/geojson/NUTS_RG_10M_2021_4326.geojson"
nuts = gpd.read_file(url)
flanders = nuts[nuts['NUTS_ID'] == 'BE2'].to_crs(gbif_spatial.crs)

# 3. Aggregate All-Time Observations & Clip
# Group strictly by specieskey
species_totals = gbif_spatial.groupby(['grid_idx', 'geometry', 'specieskey'])['observation_count'].sum().reset_index()
species_totals = gpd.GeoDataFrame(species_totals, geometry='geometry', crs=gbif_spatial.crs)
species_totals_flanders = gpd.clip(species_totals, flanders)

# 4. Plotting a Map for Each Species
species_list = species_totals_flanders['specieskey'].unique()

fig, axes = plt.subplots(nrows=len(species_list), ncols=1, figsize=(12, 6 * len(species_list)))

if len(species_list) == 1:
    axes = [axes]

for ax, species_key in zip(axes, species_list):
    species_data = species_totals_flanders[species_totals_flanders['specieskey'] == species_key]
    
    flanders.plot(ax=ax, facecolor='lightgrey', edgecolor='black', linewidth=1.5)
    
    species_data.plot(
        ax=ax, 
        column='observation_count', 
        cmap='viridis', 
        legend=True,
        legend_kwds={'label': 'Total Observation Count (All-Time)', 'shrink': 0.7},
        alpha=0.9,
        edgecolor='none'
    )
    
    ax.set_title(f"Spatial Distribution (Species Key: {species_key})", fontsize=14, fontweight='bold')
    ax.set_xlabel("Easting (m)")
    ax.set_ylabel("Northing (m)")
    ax.grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```
